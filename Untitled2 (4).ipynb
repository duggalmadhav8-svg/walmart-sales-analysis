```python
sales = walmart_sales_df['Weekly_Sales']
Q1 = np.percentile(sales, 25)
Q3 = np.percentile(sales, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(lower_bound)
print(upper_bound)
```

```python
average_sales = np.mean(walmart_sales_df['Weekly_Sales'])
print(average_sales)
```

```python
walmart_sales_df['Sales_Category'] = np.where(
    walmart_sales_df['Weekly_Sales'] > np.mean(walmart_sales_df['Weekly_Sales']),
    'Above Average',
    'Below Average'
)
print(walmart_sales_df[['Weekly_Sales', 'Sales_Category']].head())
```

```python
sales = walmart_sales_df['Weekly_Sales'].astype(float)
avg = np.mean(sales)
walmart_sales_df['Sales_Category'] = np.where(
    sales > avg,
    'Above Average',
    'Below Average'
)
print(walmart_sales_df['Sales_Category'].value_counts())
```

```python
total_sales = np.sum(walmart_sales_df['Weekly_Sales'])
walmart_sales_df['Sales_Percentage'] = (
        walmart_sales_df['Weekly_Sales'] / total_sales

) * 100
walmart_sales_df[['Weekly_Sales', 'Sales_Percentage']].head()
```

```python
mean_sales = np.mean(walmart_sales_df['Weekly_Sales'])
std_sales = np.std(walmart_sales_df['Weekly_Sales'])
walmart_sales_df['Sales_ZScore'] = (
    (walmart_sales_df['Weekly_Sales'] - mean_sales) / std_sales

)
walmart_sales_df[['Weekly_Sales', 'Sales_ZScore']].head()
```

```python
conditions = [
    walmart_sales_df['Weekly_Sales'] >= 1500000,
    walmart_sales_df['Weekly_Sales'] >= 1000000
]
choices = ['Very High','High']
walmart_sales_df['Sales_Level'] = np.select(
    conditions,
    choices,
    default = 'Normal'
)
print(walmart_sales_df[['Weekly_Sales', 'Sales_Level']].head())




```

```python
walmart_sales_df = walmart_sales_df.sort_values('Date')
plt.figure(figsize = (12, 5))
plt.plot(
    walmart_sales_df['Date'],
    walmart_sales_df['Weekly_Sales']
)
plt.title('Weekly Sales Trends Over Time')
plt.xlabel('Date')
plt.ylabel('Weekly_Sales')
plt.xticks(rotation = 45)
plt.grid(True)
plt.show()
```

## Seasonal Spikes: Massive sales peaks occur near late Q4 (late November to late December), approaching $3.8M+ due to holiday shopping.
## Consistent Baseline: Outside Q4 holiday surges, baseline weekly sales stay stable within the $0.5M to $2.3M range across all years.

```python
top_10_stores = (
    walmart_sales_df.groupby('Store_Number')['Weekly_Sales']
    .mean()
    .sort_values(ascending = False)
    .head(10)
)
plt.figure(figsize = (10, 5))
plt.bar(top_10_stores.index, top_10_stores.values)
plt.title('Top 10 Stores by Average Weekly Sales')
plt.xlabel('Store_Number')
plt.ylabel('Average Weekly Sales')
plt.grid(True)
plt.show()
```

## Top Average Performers: Stores 20, 4, and 14 lead with the highest average weekly sales, each averaging around $2.0M to $2.1M.
## Consistent Baseline: All top 10 stores maintain strong average weekly sales above $1.5M, showing consistent demand rather than one-off spikes.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    x = 'Holiday_Flag',
    y = 'Weekly_Sales',
    data = walmart_sales_df
)
plt.title('Weekly Sales: Holiday vs Non-Holiday')
plt.xlabel('Holiday Flag')
plt.ylabel('Weekly Sales')
plt.grid(True)
plt.show()
```

## Higher Baseline Average: Holiday weeks (⁠1⁠) show a higher median and upper quartile compared to regular non-holiday weeks (⁠0⁠).
## Extreme Seasonal Outliers: Non-holiday weeks display numerous extreme upper outliers (reaching up to ~$3.8M), representing major shopping periods that fall outside official holiday dates.

```python
correlation = walmart_sales_df.corr(numeric_only=True)
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot = True,
    cmap = 'coolwarm',
    fmt = '.2f'
)
plt.title('Correlation Heatmap')
plt.show()
```

## Weak External Drivers: Macro factors like Temperature (-0.06), Fuel Price (0.01), and CPI (-0.07) show virtually no linear correlation with Weekly Sales.
## Negative Store Correlation: Store Number has a moderate negative correlation (-0.34) with sales, indicating that lower store IDs represent higher-performing locations.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    x = 'Temperature',
    y = 'Weekly_Sales',
    hue = 'Holiday_Flag',
    data = walmart_sales_df
)
plt.title('Temperature vs Weekly Sales')
plt.xlabel('Temperature')
plt.ylabel('Weekly Sales')
plt.grid(True)
plt.show()
```

## Holiday Clustering: Holiday weeks (⁠1⁠, orange points) concentrate heavily in moderate-to-cool temperatures (30°F to 60°F), aligning with late Q4 holidays.
## Peak Spikes: Top sales outliers ($3.0M+) occur almost exclusively in non-holiday and holiday periods within the 30°F–60°F range.

```python
plt.figure(figsize=(8, 5))
sns.histplot(
    data = walmart_sales_df,
    x = 'Weekly_Sales',
    bins = 30,
    kde = True
)
plt.title('Distribution of Weekly Sales')
plt.xlabel('Weekly Sales')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

## Multi-Peak Pattern: The KDE curve reveals distinct sales tiers, with major volume peaks around $0.6M, $1.0M, and $1.4M.
## Skewed Distribution: The distribution remains right-skewed, dropping off sharply above $2.2M with long tail outliers up to $3.8M.

```python
plt.figure(figsize=(8, 5))
sns.countplot(
    x = 'Holiday_Flag',
    data = walmart_sales_df
)
plt.title('Number of Holiday vs Non Holiday Weeks')
plt.xlabel('Holiday Flag')
plt.ylabel('Count')
plt.grid(True)
plt.show()
```

## Class Imbalance: Non-holiday weeks (⁠0⁠) make up the vast majority of dataset records (~6,000 counts), compared to regular holiday weeks (⁠1⁠) (~450 counts).
## Data Bias: Total revenue for non-holiday weeks is much higher due to sheer volume, making average or median sales better metrics for holiday comparisons than overall sum.

```python
plt.figure(figsize=(10, 6))
correlation = walmart_sales_df.corr(numeric_only= True)
sns.heatmap(
    correlation,
    annot = True,
    cmap = 'coolwarm',
    fmt = '.2f'
)
plt.title('Correlation Heatmap of   Walmart Sales Data')
plt.show()
```

## Strong Derived Metric Tie: ⁠Weekly_Sales⁠, ⁠Sales_Percentage⁠, and ⁠Sales_ZScore⁠ show a perfect positive correlation (1.00), as they are direct linear transformations of sales.
## Low Economic Sensitivity: Factors like Unemployment (-0.11), CPI (-0.07), and Temperature (-0.06) show very weak negative correlations, indicating macroeconomic variables don't directly drive weekly sales fluctuations.
